### Prototyping PaddleOCR for text extraction

```python
#Ground truth
import pandas as pd
df = pd.read_csv("sample_essay.csv")

#OCR text extractor
#from PaddleOCR ReadME file
from paddleocr import PaddleOCR

ocr = PaddleOCR(
    text_detection_model_name="PP-OCRv6_medium_det",
    text_recognition_model_name="PP-OCRv6_medium_rec",
    engine="transformers",
    use_doc_orientation_classify=False,
    use_doc_unwarping=False,
    use_textline_orientation=False,
)

#Feature Extractor 
import nltk
nltk.download('punkt_tab')

def feature_extractor(essay):
    # Count sentences using  tokenizer
    sentences = nltk.tokenize.sent_tokenize(essay)
    sentence_count = len(sentences)
    
    # Count words 
    words = nltk.tokenize.word_tokenize(essay)
    
    # Filter out punctuation so only words remain
    words = [word for word in words if word.isalnum()]
    word_count = len(words)
    
    print(f"Words: {word_count}")
    print(f"Sentences: {sentence_count}")


#Clean Text
import re

def clean(text):
    text = text.lower()
    text = re.sub(r"[^\w\s]", "", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text


#Return cleaned text
def get_cleaned_text (extracted_text):
    
    #Most of the code here copied from ChapGPT and teaked
    
    # Filter texts with a confidence score > 0.65 
    threshold = 0.65
    filtered_texts = [
        text for text, score in zip(result[0]['rec_texts'], result[0]['rec_scores'])
        if score > threshold
    ]
    
    all_text = "\n".join(filtered_texts)

    return clean(all_text)
    
    # with open('extracted_text_high_confidence.txt', 'w', encoding='utf-8') as f:
    #     f.write(all_text)

```
